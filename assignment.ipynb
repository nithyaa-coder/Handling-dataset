```python
import pandas as pd

df=pd.read_csv("student_scores.csv")
df.info()
```

```python
df.describe()
```

```python
df["Hours"]=df["Hours"].fillna(df["Hours"].mean())
df["Scores"]=df["Scores"].fillna(df["Scores"].median())
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder

df["Result"] = df["Scores"].apply(lambda x: "Pass" if x >= 40 else "Fail")
df["Grade"] = pd.cut(df["Scores"],
                     bins=[0, 40, 60, 80, 100],
                     labels=["D", "C", "B", "A"],
                     include_lowest=True).astype(str)

le = LabelEncoder()
df["Result_Label"] = le.fit_transform(df["Result"])
print(df[["Result", "Result_Label"]].head())


```

```python
ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[["Grade"]])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(["Grade"]))

df_encoded = pd.concat([df, encoded_df], axis=1)
print(df_encoded.head())

```
